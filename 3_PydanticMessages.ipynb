```python
from langchain_ollama import ChatOllama
from pydantic import BaseModel, EmailStr

model = ChatOllama(
    model="llama3.1",
    validate_model_on_init=True,
    temperature=0.8,
    num_predict=256,
)

class UserProfile(BaseModel):
    id: int                       # Required field
    username: str                 # Required field
    email: EmailStr               # Validates standard email formats
    is_active: bool = True        # Optional field with a default value

# 2. Pass data (e.g., from an API endpoint or JSON payload)
user_data = {
    "id": "123",                  # Passed as string, automatically coerced to int
    "username": "alice_dev",
    "email": "alice@example.com"
}

# 3. Instantiate and parse
user = UserProfile(**user_data)

print(user.id)        # Output: 123 (Successfully converted to an integer!)
print(user.is_active) # Output: True (Fell back to the default value)
```

```python
structuredquestion = model.with_structured_output(UserProfile)

response = structuredquestion.invoke(
    "Generate a user profile for a new developer named Bob, with email with trustage.com domain"
)

print(response)
```
