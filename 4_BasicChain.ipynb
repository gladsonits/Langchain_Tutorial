**CHAINS**

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
```

```python
#Prompt template for the chain

prompt_template = ChatPromptTemplate.from_template(
    "You are a helpful assistant that translates {input_language} to {output_language}."
)

input_language = "English"
output_language = "French"

ready_input = prompt_template.format_messages(input_language=input_language, output_language=output_language)
```

```python
# LLM initialization
model = ChatOllama(
    model="llama3.1",
    validate_model_on_init=True,
    temperature=0.8,
    num_predict=256,
    # other params ...
)
```

```python
#String output parser initialization

output_parser = StrOutputParser()

response = model.invoke(ready_input).content

response
```

```python
# CHAIN CALLING

chain = prompt_template | model | output_parser

chain.invoke({"input_language": "English", "output_language": "French"})
```

```python
from langchain_core.runnables import RunnableSequence

chain2 = RunnableSequence(prompt_template, model, output_parser)

chain2.invoke({"input_language": "English", "output_language": "French"})
```
