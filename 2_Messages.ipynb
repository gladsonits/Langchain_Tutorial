```python
from langchain_core.prompts import PromptTemplate

user_input = input("Enter a topic for a joke: ")

# Instantiation using from_template (recommended)
prompt = PromptTemplate.from_template("Tell me a joke about {foo}")
finalprompt =prompt.format(foo= user_input)

model.invoke(finalprompt).content
```

```python
from langchain_core.prompts import ChatPromptTemplate

user_input = input("Ask something: ")

template = ChatPromptTemplate(
    [
        ("system", "You are a helpful AI bot. Your name is Carl."),
        ("human", "{user_input}"),
    ]
)

prompt_value = template.invoke({"user_input": user_input})

model.invoke(prompt_value.messages).content

# Equivalent to
# prompt_value = template.invoke({"user_input": "Hello, there!"})

# Output:
#  ChatPromptValue(
#     messages=[
#         SystemMessage(content='You are a helpful AI bot. Your name is Carl.'),
#         HumanMessage(content='Hello, there!'),
#     ]
# )
```
